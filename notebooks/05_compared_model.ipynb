```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

PROJECT = Path.cwd().parents[0]
OUTPUT_CSV = PROJECT / "results" / "data"
PLOT_DIR = OUTPUT_CSV / "plots"
PLOT_DIR.mkdir(parents=True, exist_ok=True)

scores_file = OUTPUT_CSV / "model_scores.csv"

if not scores_file.exists():
    raise FileNotFoundError(f"Không tìm thấy file metric: {scores_file}. Hãy chạy các file mô hình trước.")

df = pd.read_csv(scores_file)
print("📊 Tất cả metric đã ghi:")
print(df)

models = sorted(df["model"].unique())
splits = ["Train", "Valid", "Test"]

def get_metric(df, model, split, metric):
    rows = df[(df["model"] == model) & (df["split"] == split)]
    if len(rows) == 0:
        return np.nan
    return float(rows.iloc[0][metric])

def plot_metric(metric_name: str, filename: str, title: str):
    x = np.arange(len(models))
    width = 0.25

    plt.figure(figsize=(10, 6))

    bar_groups = []  # lưu các bar để ghi số

    for i, split in enumerate(splits):
        values = [get_metric(df, m, split, metric_name) for m in models]
        bars = plt.bar(x + i * width, values, width, label=split)
        bar_groups.append((bars, values))

    # Gắn nhãn số lên từng cột
    for bars, values in bar_groups:
        for bar, val in zip(bars, values):
            if np.isnan(val):
                continue
            height = bar.get_height()
            plt.text(
                bar.get_x() + bar.get_width() / 2,
                height,
                f"{val:.3f}",
                ha="center",
                va="bottom",
                fontsize=8,
                rotation=0,
            )

    plt.xticks(x + width, models, rotation=15)
    plt.ylabel(metric_name.capitalize())
    plt.title(title)
    plt.legend()
    plt.tight_layout()

    out_path = PLOT_DIR / filename
    plt.savefig(out_path, dpi=150)
    plt.close()
    print(f"✅ Đã lưu biểu đồ {metric_name} (Train/Valid/Test) tại: {out_path}")

# Ảnh 1: Accuracy
plot_metric(
    metric_name="accuracy",
    filename="accuracy_train_valid_test.png",
    title="So sánh Accuracy trên Train / Valid / Test cho từng mô hình",
)

# Ảnh 2: Precision
plot_metric(
    metric_name="precision",
    filename="precision_train_valid_test.png",
    title="So sánh Precision trên Train / Valid / Test cho từng mô hình",
)

# Ảnh 3: Recall
plot_metric(
    metric_name="recall",
    filename="recall_train_valid_test.png",
    title="So sánh Recall trên Train / Valid / Test cho từng mô hình",
)

# Ảnh 4: F1-score
plot_metric(
    metric_name="f1",
    filename="f1_train_valid_test.png",
    title="So sánh F1-score trên Train / Valid / Test cho từng mô hình",
)

print("\n🎉 Hoàn tất vẽ 4 biểu đồ so sánh các mô hình.")
```
